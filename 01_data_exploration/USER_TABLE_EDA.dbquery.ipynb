```sql
use databricks_wanderbricks_dataset_dais_2025.wanderbricks;

show tables;
 
--- Exploring Users Tables

-- Basic things to check while doing EDA

---1.What does the table contain?
---2.What does one row represent?
---3.What columns are available?
---4.How many  rows are there?

select * from users limit 10;

describe users;

-- Count of users

select count(*) from users;

-- Users table

-- 1.Contains information about  Wanderbricks users.
-- 2.One row represents one user
-- 3.Columns tells us ehat information we know about each user.
-- 4. There are 100000  unique users in the table.
-- 5. Users can be either individual or business.
-- 6.Users can be either individual or business
-- 7.We can study user data by country, account type and signup data.

-- What is the distribution of users by country?
select country,count(*) as num_users from users group by country order by num_users desc;

-- Distribution by user type
select user_type, count(*)as total_users
from users
group by user_type
order by total_users desc;

-- Any missing values ?
select 
Sum(case when user_id is null then 1 else 0 end) as missing_user_id,
Sum(case when email is null then 1 else 0 end) as missing_email,
Sum(case when name is null then 1 else 0 end) as missing_name,
Sum(case when Country is null then 1 else 0 end) as missing_country,
Sum(case when user_type is null then 1 else 0 end) as missing_user_type,
Sum(case when created_at is null then 1 else 0 end) as missing_created_at,
Sum(case when is_business is null then 1 else 0 end) as missing_is_business,
Sum(case when company_name is null then 1 else 0 end) as missing_company_name
from users;




-- What does our user base looks like?
select country, user_type, count(*)
from users
group by country, user_type
order by country;

-- When did users join the platform. (Earliest sign up, latest signup, users joining by month/year)
--- Earliest and latest signup
select min(created_at) as earliest_signup,
 max(created_at) as latest_signup 
from users;

--- users joining by year/month.
select year(created_at) as signup_yr,month(created_at) as signup_month, count(*) as total_users
from users
group by year(created_at),month(created_at)
order by signup_yr,signup_month;
```
